```python
from service import ApiService, ScraperService, ConversorService
from model import ChannelModel, VideoModel, PlaylistModel, SearchModel, TranscriptModel
```

```python
api = ApiService()
scraper = ScraperService()
conversor = ConversorService()
```

```python
canal_name = "@renatocariani"
canal_dados = api.get_channel(canal_name)
canal = conversor.obter_dados(canal_dados, ChannelModel)
print(canal)
```

```python
playlist_videos = api.get_playlist_videos(canal_id=canal.items[0].canal_id, limite=5)
videos_id = conversor.obter_dados(playlist_videos, PlaylistModel)
print(videos_id)
```

```python
search_dados = api.search_videos(canal_id=canal.items[0].canal_id, limite=5)
videos_id = conversor.obter_dados(search_dados, SearchModel)
print(videos_id)
```

```python
ids = [item.id.video_id for item in videos_id.items]
videos_dados = api.get_videos(ids)
videos = conversor.obter_dados(videos_dados, VideoModel)
print(videos)
```
