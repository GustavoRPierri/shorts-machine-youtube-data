## ScraperService — Transcrição e Comentários (yt-dlp)
Usa o mesmo vídeo já carregado acima via ApiService.

```python
from service import ScraperService

scraper = ScraperService(url="")

# Usa o primeiro vídeo buscado pela ApiService acima
video_url = videos[0].url if videos else "https://www.youtube.com/watch?v=jNQXAC9IVRw"
```

### Transcrição

```python
transcricao = scraper.get_transcricao(video_url, langs=["pt", "en"])

if transcricao is None:
    print("Sem transcricao disponivel para este video.")
else:
    eventos = transcricao.get("events", [])
    linhas = []
    for ev in eventos:
        for seg in ev.get("segs") or []:
            texto = seg.get("utf8", "").strip()
            if texto and texto != "\n":
                linhas.append(texto)
    print("Trecho da transcricao:")
    print(" ".join(linhas[:50]))
```

```python
# Estrutura bruta do json3 (para referencia)
if transcricao:
    print("Chaves:", list(transcricao.keys()))
    primeiro = transcricao["events"][0] if transcricao.get("events") else None
    print("Primeiro evento:", primeiro)
```

```python
# Transcrição com intervalos de tempo (tStartMs + dDurationMs do json3)
def ms_para_tempo(ms: int) -> str:
    total_seg = ms // 1000
    horas = total_seg // 3600
    minutos = (total_seg % 3600) // 60
    segundos = total_seg % 60
    if horas:
        return f"{horas}:{minutos:02d}:{segundos:02d}"
    return f"{minutos}:{segundos:02d}"

if transcricao:
    for ev in transcricao.get("events") or []:
        inicio_ms = ev.get("tStartMs", 0)
        duracao_ms = ev.get("dDurationMs", 0)
        fim_ms = inicio_ms + duracao_ms
        texto = "".join(seg.get("utf8", "") for seg in ev.get("segs") or []).strip()
        if texto and texto != "\n":
            print(f"[{ms_para_tempo(inicio_ms)} - {ms_para_tempo(fim_ms)}] {texto}")
```
