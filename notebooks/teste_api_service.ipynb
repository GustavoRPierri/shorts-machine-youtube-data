```python
import os
from dotenv import load_dotenv
from service import YoutubeService, ScraperService, ConversorService
from model import ChannelModel, VideoModel, PlaylistModel, SearchModel, CommentModel, TranscriptModel
```

```python
load_dotenv()
API_KEY =  os.getenv("YOUTUBE_API_KEY")
END_PROXY = os.getenv("ENDPOINT_PROXY")
```

```python
api = YoutubeService(key=API_KEY)
scraper = ScraperService(end_proxy=END_PROXY)
conversor = ConversorService()
```

```python
canal_name = "@renatocariani"
canal_dados = api.get_channel(canal_name)
canal = conversor.obter_dados(canal_dados, ChannelModel)
```

```python
for page in api.get_playlist(playlist_id="UUPX0gLduKAfgr-HJENa7CFw",  limite_por_pagina=5, max_paginas=1):
    playlist = conversor.obter_dados(page, PlaylistModel)
```

```python
ids = [item.conteudo.video_id for item in playlist.items]
videos_dados = api.get_videos(ids)
for page in videos_dados:
    videos = conversor.obter_dados(page, VideoModel)
```

```python
urls = [item.url for item in videos.items]
comentarios = scraper.get_comentarios(video_url=urls[3], max_comentarios=5)
conversor.obter_lista(comentarios, CommentModel)
```

```python
transcricoes = scraper.get_transcricao(video_id=ids[3], langs=["pt"])
conversor.obter_lista(transcricoes, TranscriptModel)
```
